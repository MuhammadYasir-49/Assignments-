**Q1:** Aik `Student` class banayein jismein `name` aur `roll_number` private variables (`__`) hon. In variables ko access aur modify karne ke liye getter aur setter methods likhein.

**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
# Write your code here for Q2
class BankAccount:
    def __init__(self,balance):
        self.__balance = balance

    def deposit(self,d_amount):
        self.__balance += d_amount
        print(f"Deposited: {d_amount}")

    def withdraw(self,w_amount):
        if w_amount > self.__balance:
            print(f"This much amount is not your account, Current Balance is: {self.__balance}")
        else:
            print("Done")


p1 = BankAccount(10000)
p1.deposit(5000)
p1.withdraw(20000)
```

**Q7:** Multilevel Inheritance ka use karein. Class `Animal` -> Class `Mammal` -> Class `Dog`. Dog class ke object se parent classes ke features ko access karein.

```python
# Write your code here for Q7
class Animal:
    def __init__(self,name):
        self.name = name

    def eat(self):
        print(f"{self.name} is eating.")


class Mammal(Animal):
    def __init__(self,name):
        super().__init__(name)

    def sleep(self):
        print(f"{self.name} is sleeping")


class Dog(Mammal):
    def __init__(self,name,breed):
        super().__init__(name)
        self.breed = breed
     
    def bark(self):
        print(f"{self.name} ({self.breed}) is barking")


a1 = Dog("tommy", "pandu")

a1.eat()
a1.sleep()
a1.bark()
```

**Q8:** Multiple Inheritance ka istemal karte hue do parent classes `Father` aur `Mother` banayein, aur unse aik child class `Child` derive karein jo dono ke habits inherit kare.

```python
# Write your code here for Q8
class Father:
    def __init__(self):
        super().__init__()
        self.father_habits = ["Watching Football", "Waking up early morning"]

    def show_father_habits(self):
        for i in self.father_habits:
            print(i)

class Mother:
    def __init__(self):
        super().__init__()
        self.mother_habits = ["Cooking food", "Reading books"]

    def show_mother_habits(self):
        for i in self.mother_habits:
            print(i)

class Child(Father,Mother):
    def __init__(self):
        super().__init__()

    def show_child_habits(self):
        print(f"Child got these habits from his Father and Mother")
        self.show_father_habits()
        print()
        self.show_mother_habits()

child = Child()
child.show_child_habits()
```

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
# Write your code here for Q10
class Shape:
    def show(self):
        print("I'm a shape")


class Circle(Shape):
    def area(self, r):
        print("Area of circle =", 3.14 * r * r)


class Square(Shape):
    def area(self, s):
        print("Area of square =", s * s)


c = Circle()
c.show()
c.area(5)

sq = Square()
sq.show()
sq.area(4)
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
# Write your code here for Q13
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Number(self.value + other.value)


n1 = Number(10)
n2 = Number(20)

n3 = n1 + n2

print("n1 =", n1.value)
print("n2 =", n2.value)
print("n1 + n2 =", n3.value)
```

**Q17:** Q16 wali `Computer` class ko inherit karke do child classes `Laptop` aur `Desktop` banayein aur dono mein `process()` method ko mandatory implement karein.

```python
# Write your code here for Q17
from abc import ABC, abstractmethod


class Computer(ABC):
    @abstractmethod
    def process(self):
        pass


class Laptop(Computer):
    def process(self):
        print("Laptop is processing data")


class Desktop(Computer):
    def process(self):
        print("Desktop is processing data")


l = Laptop()
d = Desktop()

l.process()
d.process()
```

**Q18:** Aik abstract class `Car` banayein jisme aik concrete (normal) method `start_engine()` ho aur aik abstract method `drive()` ho. Child class bana kar dono ko test karein.

```python
# Write your code here for Q18
from abc import ABC, abstractmethod


class Car(ABC):
    def start_engine(self):
        print("Engine started")

    @abstractmethod
    def drive(self):
        pass


class Honda(Car):
    def drive(self):
        print("You are driving the car")


h = Honda()
h.start_engine()
h.drive()
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
# Write your code here for Q20
from abc import ABC, abstractmethod


class Payment(ABC):
    def show_amount(self, amount):
        print("Total amount =", amount, "rupay")

    @abstractmethod
    def pay(self, amount):
        pass


class CreditCardPayment(Payment):
    def pay(self, amount):
        self.show_amount(amount)
        print("Done")


class EasypaisaPayment(Payment):
    def pay(self, amount):
        self.show_amount(amount)
        print("Easypaisa payment Done")


card = CreditCardPayment()
easypaisa = EasypaisaPayment()

card.pay(5000)
print()
easypaisa.pay(2000)
```
